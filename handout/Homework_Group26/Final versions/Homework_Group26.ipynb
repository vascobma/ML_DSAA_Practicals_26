#### Helper Functions

We define two helper functions that will be used later in the EDA:

- **`missing_table()`** — Shows the number and percentage of missing values for each column, helping us identify which variables have missing data.
- **`normalize_string_columns()`** — Removes extra spaces and accents, and converts text to lowercase. This helps us check whether inconsistent spelling creates duplicate categories.

We define these functions here to keep the code organized and use them in the relevant sections of the analysis.

```python
def missing_table(df):
    """Summarize missing values for each column."""
    n = df.isna().sum()
    pct = (df.isna().mean() * 100).round(1)
    out = pd.DataFrame({"n_missing": n, "%": pct})
    return out[out["n_missing"] > 0].sort_values(
        "n_missing", ascending=False
    )
```

```python
def normalize_string_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Standardize text columns and remove accents and extra spaces."""
    df_clean = df.copy()
    df_clean.columns = df_clean.columns.str.strip().str.lower()
    text_columns = df_clean.select_dtypes(include=["object", "string"]).columns

    for col in text_columns:
        df_clean[col] = df_clean[col].astype(str).str.strip().str.lower()
        df_clean[col] = df_clean[col].apply(
            lambda x: "".join(
                c for c in unicodedata.normalize("NFD", x)
                if unicodedata.category(c) != "Mn"
            ) if pd.notna(x) else x
        )
        df_clean[col] = df_clean[col].replace("nan", pd.NA)

    return df_clean
```

#### Preprocessing Helper Functions

These functions are defined here and used in Section 2. They handle missing purchase history, correct anomalous values, join selected customer information, and create purchase features. They are applied in the same way to training and test data; statistical preprocessing is fitted later using training data only.

```python
def clean_missing_history(df):
    """Fill structurally missing purchase-history values and add missingness flags."""
    df = df.copy()

    # Record missing standard deviations
    df["std_purchase_missing"] = df["prior_std_purchase_value"].isna().astype(int)
    df["std_gap_missing"] = df["prior_std_gap_days"].isna().astype(int)

    # Fill undefined standard deviations with zero
    df.loc[
        (df["prior_purchase_count"] < 2) &
        df["prior_std_purchase_value"].isna(),
        "prior_std_purchase_value"
    ] = 0

    df.loc[
        (df["prior_purchase_count"] < 3) &
        df["prior_std_gap_days"].isna(),
        "prior_std_gap_days"
    ] = 0

    # Handle customers with no previous purchases
    no_history = df["prior_purchase_count"] == 0
    df["no_previous_purchases"] = no_history.astype(int)

    history_cols = [
        "days_since_previous_purchase",
        "prior_mean_purchase_value",
        "prior_min_purchase_value",
        "prior_max_purchase_value"
    ]

    df.loc[no_history, history_cols] = (
        df.loc[no_history, history_cols].fillna(0)
    )

    # Handle insufficient history for average purchase gaps
    limited_history = df["prior_purchase_count"] < 2
    df["insufficient_gap_history"] = limited_history.astype(int)

    df.loc[
        limited_history & df["prior_mean_gap_days"].isna(),
        "prior_mean_gap_days"
    ] = 0

    return df


def clean_anomalies(df):
    """Correct invalid historical gaps, annual spending and extreme placeholders."""
    df = df.copy()

    # Recalculate average purchase values
    expected = (
        df["prior_total_spend"] /
        df["prior_purchase_count"].replace(0, np.nan)
    )

    valid = expected.notna()
    df.loc[valid, "prior_mean_purchase_value"] = expected[valid]

    # Identify negative annual spending
    previous_flag = df.get("spend_365d_negative", 0)

    df["spend_365d_negative"] = (
        (previous_flag == 1) | (df["spend_365d"] < 0)
    ).astype(int)

    # Replace negative time intervals and annual spending with NaN
    for col in [
        "days_since_previous_purchase",
        "prior_mean_gap_days",
        "spend_365d"
    ]:
        df.loc[df[col] < 0, col] = np.nan

    # Replace suspicious values of 1 billion
    for col in ["prior_mean_purchase_value", "spend_365d"]:
        df.loc[df[col] == 1_000_000_000, col] = np.nan

    
    # Correct Inconsistent Maximum Previous Purchases

    invalid_max = (
        (df["prior_purchase_count"] > 0)
        & df["prior_max_purchase_value"].notna()
        & df["prior_total_spend"].notna()
        & (
            df["prior_max_purchase_value"]
            > df["prior_total_spend"] + 0.01
        )
    )

    df.loc[invalid_max, "prior_max_purchase_value"] = np.nan

    return df


def merge_customer_info(df, customer_table):
    """Add two customer attributes without changing the number of purchases."""
    lookup = customer_table[["customer_id", "commercial_region", "customer_since_year"]]
    result = df.merge(lookup, on="customer_id", how="left", validate="many_to_one")
    assert len(result) == len(df), "The customer merge changed the number of purchases."
    return result



def add_date_features(df):
    """Create categorical month and weekday variables from the purchase date."""
    df = df.copy()

    df["purchase_date"] = pd.to_datetime(
        df["purchase_date"], errors="coerce"
    )

    df["purchase_month"] = (
        df["purchase_date"].dt.month.astype("Int64").astype(str)
    )

    df["purchase_weekday"] = (
        df["purchase_date"].dt.dayofweek.astype("Int64").astype(str)
    )

    return df


def add_purchase_features(df):
    """Add gap and recent-spending ratios using recorded purchase history."""
    df = df.copy()

    # Compare current purchase gap with usual purchase gap
    df["purchase_gap_ratio"] = (
        df["days_since_previous_purchase"] /
        df["prior_mean_gap_days"].replace(0, np.nan)
    )

    # Calculate share of annual spending in the last 30 days
    df["recent_spend_share"] = (
        df["spend_30d"] /
        df["spend_365d"].replace(0, np.nan)
    )

    return df


def values_are_finite(matrix):
    """Verify that a dense or sparse processed matrix contains no NaN or infinities."""
    values = matrix.data if hasattr(matrix, "toarray") else matrix
    return bool(np.isfinite(values).all())
```

#### 1.7.4. Numerical consistency checks

We also want to check whether related numerical variables are consistent with each other.

For example, transaction amounts should follow the expected calculations, and purchase counts over longer periods should be at least as large as those over shorter periods.

These checks can help us identify possible errors that were not detected by simply looking at minimum and maximum values.

```python
# • Check Numerical Consistency

# Exclude completely empty purchase observations
d = train.dropna(subset=["Total a Pagar"])

# Check transaction calculations
check_total = np.isclose(
    d["Total a Pagar"],
    d["Total Liquido"] + d["Total Imp_"],
    atol=0.05
)

check_net = np.isclose(
    d["Total Liquido"],
    d["Total Bruto"] - d["Total Desc_ Linha"] - d["Total Desc_ Global"],
    atol=0.05
)

# Check purchase counts across time windows
check_counts = (
    (d["purchase_count_30d"] <= d["purchase_count_90d"]) &
    (d["purchase_count_90d"] <= d["purchase_count_180d"]) &
    (d["purchase_count_180d"] <= d["purchase_count_365d"])
)

# Check spending across time windows where values are available
spend_data = d.dropna(subset=["spend_180d", "spend_365d"])

check_spend = (
    spend_data["spend_180d"] <= spend_data["spend_365d"] + 0.01
)

# Check Maximum Previous Purchase Against Total Spending

check_max = (
    (d["prior_max_purchase_value"] <= d["prior_total_spend"] + 0.01)
    | d["prior_max_purchase_value"].isna()
)


# Summarize the results

consistency = pd.DataFrame({
    "Rule": [
        "Amount paid = Net amount + Taxes",
        "Net amount = Gross amount - Discounts",
        "Purchase counts: 30d <= 90d <= 180d <= 365d",
        "Spending: 180d <= 365d",
        "Prior max <= prior total spend"
    ],
    "Rows Checked": [
        len(d), len(d), len(d), len(spend_data), len(d)
    ],
    "Violations": [
        (~check_total).sum(),
        (~check_net).sum(),
        (~check_counts).sum(),
        (~check_spend).sum(),
        (~check_max).sum()
    ]
})

display(consistency)
```

**What we see:**

The transaction calculations and purchase counts are consistent.

However, we found 15 inconsistencies in historical spending and 45 cases where the maximum previous purchase exceeds the total historical spending.

We will investigate these values during preprocessing.

### 2.5. Chronological Training and Validation Split

Test observations come from a later period, so we validate on the **most recent 20% of dated training observations** rather than selecting rows randomly. We also exclude purchases within the **90 days before the validation period** to reduce overlap between training labels and the validation window. Observations without a valid purchase date are excluded from this split. Identifiers, the date, target and `random_noise` are not model features.

```python
# Create Time-Aware Validation Split

data = train_clean.dropna(subset=["purchase_date"]).copy()
cutoff = data["purchase_date"].quantile(0.80)
train_part = data[data["purchase_date"] < cutoff - pd.Timedelta(days=90)]
val_part = data[data["purchase_date"] >= cutoff]

drop_cols = ["repeat_purchase_90d", "ID", "customer_id", "purchase_date", "random_noise"]

X_train = train_part.drop(columns=drop_cols)
X_val = val_part.drop(columns=drop_cols)
y_train = train_part["repeat_purchase_90d"]
y_val = val_part["repeat_purchase_90d"]

print("Training observations:", len(X_train))
print("Validation observations:", len(X_val))
print("Purged observations:", len(data) - len(train_part) - len(val_part))
print("Latest training date:", train_part["purchase_date"].max().date())
print("First validation date:", val_part["purchase_date"].min().date())

assert train_part["purchase_date"].max() < cutoff - pd.Timedelta(days=90)
assert val_part["purchase_date"].min() >= cutoff
```

**What we see:**

- We obtained 5,024 training observations and 1,298 validation observations.
- We excluded 147 observations to maintain the 90-day gap between training and validation.
- The latest training purchase is from March 18, 2023, while validation starts on June 17, 2023.

This split allows us to evaluate the model on later purchases while reducing the risk of data leakage.
The 65 observations without valid purchase dates were also excluded, since they cannot be placed in the chronological split.

### 2.6. Fitting the Preprocessing Pipeline

For numerical variables, we fill remaining missing values using the training median and then apply `StandardScaler`.

Scaling puts numerical variables on a similar scale, which is important for models such as Logistic Regression and KNN. Tree-based models generally do not require scaling.

For categorical variables, we fill missing values with `Missing` and use One-Hot Encoding, grouping categories that appear fewer than 10 times in training. The one-hot encoded variables are not scaled.

The preprocessing pipeline is fitted only on the training data and then applied to validation and test data to avoid data leakage.

```python
# Fit Preprocessing Pipeline on Training Data Only

numerical_cols = X_train.select_dtypes(include="number").columns.tolist()
categorical_cols = X_train.select_dtypes(include=["object", "string"]).columns.tolist()

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value="Missing")),
    ("encoder", OneHotEncoder(handle_unknown="infrequent_if_exist", min_frequency=10))
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, numerical_cols),
    ("cat", cat_pipeline, categorical_cols)
])

X_train_processed = preprocessor.fit_transform(X_train)
X_val_processed = preprocessor.transform(X_val)
feature_names = preprocessor.get_feature_names_out()

print("Training shape:", X_train_processed.shape)
print("Validation shape:", X_val_processed.shape)
print("Output features:", len(feature_names))
```

**What we see:**

After imputation, scaling and encoding, we obtained 137 features for both training and validation.

The preprocessing pipeline was fitted only on training data and then applied to validation, keeping the transformations consistent and avoiding data leakage.

**What we see:**

The test dataset was transformed into 986 observations with the same 137 features as training and validation.

All test observations and their original IDs were preserved. No preprocessing steps were refitted on test data.

### 3.1. Feature Selection Strategy

We exclude `ID`, `customer_id` and `random_noise` from modelling, as decided in Section 2. Identifiers could lead to memorization rather than useful patterns. We also examine correlated numerical features and potentially identifier-like categorical variables.

Following the ideas considered by our group, we look at correlations **only on the training split**. This is a diagnostic check, not a separate selection method: highly correlated variables are not removed automatically, because the modelling comparisons will determine whether keeping them is useful.

```python
# Investigate Highly Correlated Numerical Features (Training Data Only)

numeric_training = X_train.select_dtypes(include="number")
corr_matrix_fs = numeric_training.corr().abs()
upper_triangle = corr_matrix_fs.where(np.triu(np.ones(corr_matrix_fs.shape), k=1).astype(bool))
correlation_pairs = upper_triangle.stack().sort_values(ascending=False)
strong_pairs = correlation_pairs[correlation_pairs > 0.85]

print("Numerical feature pairs with absolute correlation above 0.85:", len(strong_pairs))
display(strong_pairs.head(10).rename("Absolute correlation").reset_index().rename(
    columns={"level_0": "Feature 1", "level_1": "Feature 2"}
))
```

**What we see:** There are 13 strongly correlated numerical pairs in the training split, including overlapping transaction amounts and purchase counts. Two missing-history indicators are also perfectly correlated. We do not automatically remove these variables, because the selection methods will compare their usefulness for prediction.

### 3.2. Baseline and Filter Method — SelectKBest

We first evaluate a model with all features and a simple baseline that always predicts a repeat purchase. We then use **SelectKBest with the ANOVA F-score** to select the 30 highest-scoring encoded features, as discussed in Week 4.

We choose 30 as an initial feature-count setting, not an optimized value. The ANOVA selector is fitted on **training data only**.

```python
# Baseline and Full Logistic Regression

baseline_predictions = np.ones(len(y_val), dtype=int)
baseline_f1 = f1_score(y_val, baseline_predictions)
baseline_auc = roc_auc_score(y_val, baseline_predictions)

model_all = LogisticRegression(max_iter=1000, random_state=42)
model_all.fit(X_train_processed, y_train)
f1_all = f1_score(y_val, model_all.predict(X_val_processed))
auc_all = roc_auc_score(y_val, model_all.predict_proba(X_val_processed)[:, 1])

print(f"Always-positive baseline: F1={baseline_f1:.4f}, ROC-AUC={baseline_auc:.4f}")
print(f"All features ({len(feature_names)}): F1={f1_all:.4f}, ROC-AUC={auc_all:.4f}")
```

**What we see:**

SelectKBest reduces the number of features from 137 to 30.

However, ROC-AUC decreases from 0.7976 to 0.7892, while F1 remains almost unchanged (0.7524 to 0.7523).

This suggests that the smaller feature set loses some useful predictive information.

```python
# SelectKBest Filter Method

selector = SelectKBest(score_func=f_classif, k=30)
X_train_selected = selector.fit_transform(X_train_processed, y_train)
X_val_selected = selector.transform(X_val_processed)
X_test_selected = selector.transform(X_test_processed)

model_selected = LogisticRegression(max_iter=1000, random_state=42)
model_selected.fit(X_train_selected, y_train)
f1_selected = f1_score(y_val, model_selected.predict(X_val_selected))
auc_selected = roc_auc_score(y_val, model_selected.predict_proba(X_val_selected)[:, 1])

print("Selected features:", X_train_selected.shape[1])
print(f"Validation F1={f1_selected:.4f}, ROC-AUC={auc_selected:.4f}")
```

**What we see:**

- The always-positive baseline achieves an F1-score of 0.7525, but a ROC-AUC of 0.5000.
- Logistic Regression with all 137 features achieves an F1-score of 0.7524 and a ROC-AUC of 0.7976.

The high baseline F1 results from always predicting the positive class, rather than distinguishing between customers.

Although Logistic Regression provides better ranking performance, its F1-score at the default threshold of 0.5 remains similar to the baseline. We will investigate different classification thresholds during modelling.

### 3.3. Wrapper Method — RFE

**Recursive Feature Elimination (RFE)** fits Logistic Regression repeatedly and removes the least important features based on the estimated coefficients.

We ask RFE for 30 features, matching SelectKBest, so the two methods can be compared using the same feature count.

```python
# Wrapper Feature Selection — RFE

rfe_selector = RFE(
    estimator=LogisticRegression(max_iter=1000, random_state=42),
    n_features_to_select=30,
    step=5
)

X_train_rfe = rfe_selector.fit_transform(X_train_processed, y_train)
X_val_rfe = rfe_selector.transform(X_val_processed)
X_test_rfe = rfe_selector.transform(X_test_processed)

model_rfe = LogisticRegression(max_iter=1000, random_state=42)
model_rfe.fit(X_train_rfe, y_train)
f1_rfe = f1_score(y_val, model_rfe.predict(X_val_rfe))
auc_rfe = roc_auc_score(y_val, model_rfe.predict_proba(X_val_rfe)[:, 1])

print("Selected features:", X_train_rfe.shape[1])
print(f"Validation F1={f1_rfe:.4f}, ROC-AUC={auc_rfe:.4f}")
```

**What we see:**

RFE selects 30 features and achieves an F1-score of 0.7656 and ROC-AUC of 0.7924.

It improves F1 compared with SelectKBest and the full-feature model, although its ROC-AUC remains slightly below the full-feature model.

RFE provides the highest F1-score among the feature-selection methods tested.

### 3.4. Embedded Method — L1 Logistic Regression

L1 regularization can shrink less relevant coefficients to zero. We use `SelectFromModel` with L1 Logistic Regression to keep the features whose coefficients remain nonzero.

We choose `C=0.1` as an initial regularization setting, not an optimized value. All coefficients are estimated from the training split.

```python
# Use L1 settings compatible with the installed scikit-learn version

sklearn_version = tuple(int(part) for part in sklearn.__version__.split(".")[:2])

l1_options = {"l1_ratio": 1} if sklearn_version >= (1, 8) else {"penalty": "l1"}

embedded_selector = SelectFromModel(
    LogisticRegression(
        **l1_options,
        solver="liblinear",
        C=0.1,
        max_iter=1000,
        random_state=42
    ),
    threshold=1e-8
)

X_train_embedded = embedded_selector.fit_transform(X_train_processed, y_train)
X_val_embedded = embedded_selector.transform(X_val_processed)
X_test_embedded = embedded_selector.transform(X_test_processed)
embedded_features = feature_names[embedded_selector.get_support()]

model_embedded = LogisticRegression(max_iter=1000, random_state=42)
model_embedded.fit(X_train_embedded, y_train)
f1_embedded = f1_score(y_val, model_embedded.predict(X_val_embedded))
auc_embedded = roc_auc_score(y_val, model_embedded.predict_proba(X_val_embedded)[:, 1])

# Verify that L1 actually selected fewer features

print("scikit-learn version:", sklearn.__version__)
print("L1 settings:", l1_options)

assert X_train_embedded.shape[1] < len(feature_names), "L1 did not reduce the features."
print("Selected features:", X_train_embedded.shape[1])
print(f"Validation F1={f1_embedded:.4f}, ROC-AUC={auc_embedded:.4f}")
```

### 4.1. Assessment Strategy and Metrics

**How we assess the models.** The test set comes from a **later period** than the training set, so a random split would give optimistic results. We therefore use two levels of assessment, both **chronological**:

| Level | Data | Purpose |
|---|---|---|
| **Cross-validation inside the training part** | 5 expanding-window folds, always training on the past and validating on the next period, with a 90-day gap | choose hyperparameters and the classification threshold |
| **Validation set** | the most recent 20% of dated training purchases (June 2023 onwards), created in 2.5 | compare the final candidates **once**, without having tuned anything on it |

This way the validation set is not used to tune hyperparameters or the threshold, which reduces the optimism warned about in 3.7.

**Which metrics, and why.**

| Metric | What it measures | Why it fits this problem |
|---|---|---|
| **ROC-AUC** | how well the model *ranks* repeat purchasers above non-repeaters, for any threshold | it does not depend on a threshold and is the best summary of the model's discrimination |
| **F1-score** (class 1) | the balance between precision and recall for the repeat-purchase class | the predictions we submit are hard labels (0/1), so we also need a metric that evaluates labels. F1 is the one we use to choose the threshold |
| **Accuracy** | share of correct labels | easy to read, but misleading here: always predicting "repeat" is already right about 60% of the time |
| **Precision and recall** | how many predicted repeaters are real, and how many real repeaters we find | they explain *why* F1 changes when we move the threshold |
| **Log-loss** | quality of the predicted probabilities | detects models that rank well but are badly calibrated |

**The baseline to beat:** a model that always predicts "repeat purchase". It has an F1-score of about 0.75 (because class 1 is the majority) but a ROC-AUC of 0.5, which is why we never judge a model by F1 alone.

The classification threshold is a choice, not a given: by default 0.5, but F1 usually improves with a lower threshold. We choose it with the cross-validation predictions and the metric set in `DECISION_METRIC` (see the note below).

> **Note:** the sample submission contains 0/1 labels, so the competition score is computed on labels. If the Kaggle page (Evaluation tab) shows that the metric is accuracy instead of F1, change `DECISION_METRIC` to `"accuracy"` and run Section 4 again.

```python
# • Tools for Section 4
import itertools

from sklearn.base import clone
from sklearn.ensemble import (
    ExtraTreesClassifier, HistGradientBoostingClassifier,
    RandomForestClassifier, VotingClassifier,
)
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score, confusion_matrix, log_loss, precision_score, recall_score, roc_curve,
)
from sklearn.neighbors import KNeighborsClassifier

# Metric used to choose the classification threshold: "f1" or "accuracy"
DECISION_METRIC = "f1"


def evaluate_predictions(y_true, proba, threshold=0.5):
    """Return all assessment metrics for predicted probabilities and a threshold."""
    pred = (proba >= threshold).astype(int)
    return {
        "ROC-AUC": roc_auc_score(y_true, proba),
        "F1": f1_score(y_true, pred),
        "Accuracy": accuracy_score(y_true, pred),
        "Precision": precision_score(y_true, pred, zero_division=0),
        "Recall": recall_score(y_true, pred),
        "Log-loss": log_loss(y_true, np.clip(proba, 1e-6, 1 - 1e-6)),
    }


def best_threshold(y_true, proba, metric=DECISION_METRIC):
    """Threshold (between 0.20 and 0.80) that maximises F1 or accuracy."""
    grid = np.round(np.arange(0.20, 0.801, 0.01), 2)
    metric_fn = f1_score if metric == "f1" else accuracy_score
    scores = [metric_fn(y_true, (proba >= t).astype(int)) for t in grid]
    return float(grid[int(np.argmax(scores))]), float(max(scores))


# Baseline: always predict "repeat purchase" (threshold 0 labels everything as 1)
y_val_array = y_val.to_numpy()
y_train_array = y_train.to_numpy()
baseline_proba = np.full(len(y_val_array), y_train_array.mean())
baseline_metrics = evaluate_predictions(y_val_array, baseline_proba, threshold=0.0)

print(f"Repeat-purchase rate: training part {y_train_array.mean():.3f} | validation {y_val_array.mean():.3f}")
display(pd.DataFrame([baseline_metrics], index=["Always predict 1"]).round(4))
```

**What we see:** the repeat-purchase rate rises from 55% in the training part to 60% in the validation set, which confirms that behaviour changes over time. The baseline that always predicts "repeat purchase" reaches an F1-score of 0.7525 and an accuracy of 0.6032, but its ROC-AUC is exactly 0.5. These are the numbers every model has to beat.

### 4.3. Time-Aware Cross-Validation

To tune the models without touching the validation set, we use cross-validation **inside the training part**, but respecting time:

1. we sort the training purchases by date and divide them into 6 consecutive blocks;
2. in each of the 5 folds we **train on all the blocks before** a block and **validate on that block**;
3. we remove from training the purchases less than **90 days before** the validation block, because their target (repurchase in 90 days) overlaps with the validation period.

A normal random k-fold would mix past and future and would overestimate performance.

```python
# Sort the training part by purchase date
train_order = np.argsort(train_part["purchase_date"].to_numpy(), kind="stable")
X_cv = X_tr_f[train_order]
y_cv = y_train_array[train_order]
dates_cv = pd.DatetimeIndex(train_part["purchase_date"].to_numpy()[train_order])


def chronological_folds(dates, n_blocks=6, gap_days=90):
    """Expanding-window folds: train on the past, validate on the next block, leaving a gap."""
    edges = np.linspace(0, len(dates), n_blocks + 1).astype(int)
    folds = []
    for k in range(1, n_blocks):
        validation_idx = np.arange(edges[k], edges[k + 1])
        block_start = dates[edges[k]]
        past_idx = np.arange(0, edges[k])
        train_idx = past_idx[np.asarray(dates[past_idx] < block_start - pd.Timedelta(days=gap_days))]
        folds.append((train_idx, validation_idx))
    return folds


cv_folds = chronological_folds(dates_cv)

display(pd.DataFrame({
    "Training rows": [len(tr) for tr, va in cv_folds],
    "Validation rows": [len(va) for tr, va in cv_folds],
    "Validation starts": [dates_cv[va[0]].date() for tr, va in cv_folds],
    "Validation ends": [dates_cv[va[-1]].date() for tr, va in cv_folds],
    "Repeat rate (validation)": [round(y_cv[va].mean(), 3) for tr, va in cv_folds],
}, index=[f"Fold {i}" for i in range(1, 6)]))
```

**What we see:** the five folds have about 840 validation rows each. The training part grows from about 750 to 3,900 rows, and the validation blocks move forward in time. Each fold predicts a later period using only earlier data, as the test set will do.

### 4.4. Hyperparameter Tuning

Each model has settings (hyperparameters) that control how much it can adapt to the training data. Too much freedom makes it memorise noise (**overfitting**), and too little makes it miss real patterns. We try a **small grid** of values for each of the four models, measure the average **ROC-AUC across the 5 time-aware folds**, and keep the best combination.

We use ROC-AUC to tune because it is stable and does not depend on the threshold, which we choose afterwards.

```python
search_space = {
    "Logistic Regression": (
        LogisticRegression, dict(max_iter=3000, random_state=SEED),
        {"C": [0.03, 0.1, 0.3, 1, 3]},
    ),
    "Random Forest": (
        RandomForestClassifier, dict(n_estimators=200, n_jobs=-1, random_state=SEED),
        {"min_samples_leaf": [3, 5, 10, 20], "max_features": ["sqrt", 0.5]},
    ),
    "Extra Trees": (
        ExtraTreesClassifier, dict(n_estimators=200, n_jobs=-1, random_state=SEED),
        {"min_samples_leaf": [3, 5, 10, 20], "max_features": ["sqrt", 0.5, 1.0]},
    ),
    "Gradient Boosting": (
        HistGradientBoostingClassifier, dict(min_samples_leaf=40, random_state=SEED),
        {"learning_rate": [0.03, 0.05], "max_leaf_nodes": [7, 15],
         "max_iter": [100, 200], "l2_regularization": [1.0, 5.0]},
    ),
}


def cv_auc(model, X, y, folds):
    """Mean and standard deviation of the ROC-AUC across the chronological folds."""
    scores = []
    for train_idx, valid_idx in folds:
        fitted = clone(model).fit(X[train_idx], y[train_idx])
        scores.append(roc_auc_score(y[valid_idx], fitted.predict_proba(X[valid_idx])[:, 1]))
    return float(np.mean(scores)), float(np.std(scores))


best_params, tuning_rows = {}, []
for model_name, (model_class, fixed, grid) in search_space.items():
    results = []
    for values in itertools.product(*grid.values()):
        params = dict(zip(grid.keys(), values))
        mean_auc, std_auc = cv_auc(model_class(**fixed, **params), X_cv, y_cv, cv_folds)
        results.append((mean_auc, std_auc, params))
    results.sort(key=lambda r: r[0], reverse=True)
    best_params[model_name] = results[0][2]
    tuning_rows.append({
        "Model": model_name, "Combinations tried": len(results),
        "Best CV ROC-AUC": results[0][0], "Std across folds": results[0][1],
        "Worst CV ROC-AUC": results[-1][0], "Best settings": results[0][2],
    })

tuning_summary = pd.DataFrame(tuning_rows).set_index("Model")
display(tuning_summary.round(4))
```

**What we see:**

- All four models reach a similar cross-validated ROC-AUC (about 0.78 to 0.79). The best is **Extra Trees**, followed by Logistic Regression, Random Forest and Gradient Boosting.
- The difference between the best and the worst combination is up to 0.02 for the tree models, so tuning matters. The best settings are conservative: leaves with at least 10 purchases for the forests and strong regularisation for the boosting. This means the data are noisy and models that adapt less generalise better.
- The standard deviation across folds (about 0.015 to 0.02) is as large as the differences between models. **The models are not clearly different from each other**, which suggests combining them.

### 4.5. Model Comparison and Threshold Choice

We now have four tuned models. We also build an **ensemble** (soft voting) that averages the predicted probabilities of the four. Averaging models that make different mistakes usually reduces variance and is a reliable way to improve results.

For each candidate we:
1. obtain **out-of-fold predictions**: for every purchase in the 5 validation blocks, the probability predicted by a model that never saw it;
2. choose the **threshold** that maximises `DECISION_METRIC` on those predictions;
3. train on the whole training part and assess **once** on the validation set.

The chosen model is the one with the **best cross-validated ROC-AUC**, not the best on the validation set. This prevents us from choosing by looking at the validation set.

```python
# Final versions of the four models (more trees than during the search)
def build_model(model_name):
    model_class, fixed, _ = search_space[model_name]
    params = {**fixed, **best_params[model_name]}
    if "n_estimators" in params:
        params["n_estimators"] = 500
    return model_class(**params)


model_names = list(search_space)

# 1. Out-of-fold predictions for each model
oof_index = np.concatenate([va for tr, va in cv_folds])
y_oof = y_cv[oof_index]
oof_proba = {name: np.zeros(len(oof_index)) for name in model_names}
position = 0
for train_idx, valid_idx in cv_folds:
    for name in model_names:
        fitted = build_model(name).fit(X_cv[train_idx], y_cv[train_idx])
        oof_proba[name][position:position + len(valid_idx)] = fitted.predict_proba(X_cv[valid_idx])[:, 1]
    position += len(valid_idx)

# The ensemble averages the four probabilities
oof_proba["Ensemble (soft voting)"] = np.mean([oof_proba[name] for name in model_names], axis=0)

# 2. Validation predictions of models trained on the whole training part
val_proba = {name: build_model(name).fit(X_tr_f, y_train_array).predict_proba(X_va_f)[:, 1] for name in model_names}
val_proba["Ensemble (soft voting)"] = np.mean([val_proba[name] for name in model_names], axis=0)

# 3. Tables
cv_rows, val_rows = [], []
for name in oof_proba:
    threshold, oof_score = best_threshold(y_oof, oof_proba[name])
    cv_rows.append({"Model": name, "CV ROC-AUC": roc_auc_score(y_oof, oof_proba[name]),
                    "Threshold": threshold, f"CV {DECISION_METRIC}": oof_score})
    val_rows.append({"Model": name, "Threshold": threshold,
                     "F1 @ 0.5": f1_score(y_val_array, (val_proba[name] >= 0.5).astype(int)),
                     **evaluate_predictions(y_val_array, val_proba[name], threshold)})

cv_table = pd.DataFrame(cv_rows).set_index("Model")
val_table = pd.DataFrame(val_rows).set_index("Model")

print("Cross-validation (5 chronological folds), used to choose the model and the threshold:")
display(cv_table.round(4))
print("Validation set (assessed once):")
display(val_table.round(4))
```

**What we see:**

- **Cross-validation:** the ensemble has the highest ROC-AUC (about 0.789), slightly above Extra Trees and Logistic Regression. The optimal thresholds for F1 are **below 0.5** (about 0.41 to 0.45): since "repeat" is the majority class, F1 rewards predicting it more often. The cross-validated F1 (about 0.75) is lower than the validation F1 because the folds cover older periods, when the repeat rate was lower (50% to 59%), and so the always-repeat baseline is also lower there (F1 of about 0.71).
- **Validation:** with its CV threshold, every model beats the baseline (F1 0.7525, ROC-AUC 0.5) in discrimination. The ensemble reaches a ROC-AUC of about 0.817, the highest of all, and an F1 of about 0.79. The individual forests have a similar or slightly higher F1, but with a different precision and recall balance (they predict "repeat" more often).
- **The effect of the threshold is clear:** for all models, F1 at the chosen threshold is higher than F1 at 0.5, with a better recall and a small loss in precision.
- The differences in F1 between models (about 0.01 to 0.02) are inside the uncertainty of a validation set of 1,298 purchases, as the intervals below show.

**Decision:** we choose the **ensemble**. Its advantage is not a big difference in the score but robustness: it has the best cross-validated ROC-AUC, one of the best validation scores, and averaging four different models makes it less dependent on the particular choices of any one of them.

```python
# The chosen model is the one with the best CROSS-VALIDATED ROC-AUC (not the best on validation)
chosen_name = cv_table["CV ROC-AUC"].idxmax()
chosen_threshold = float(cv_table.loc[chosen_name, "Threshold"])
print("Chosen model:", chosen_name)
print("Chosen threshold (from cross-validation):", chosen_threshold)

# Effect of the threshold on the out-of-fold predictions
thresholds = np.round(np.arange(0.20, 0.801, 0.01), 2)
f1_curve = [f1_score(y_oof, (oof_proba[chosen_name] >= t).astype(int)) for t in thresholds]
acc_curve = [accuracy_score(y_oof, (oof_proba[chosen_name] >= t).astype(int)) for t in thresholds]

plt.figure(figsize=(7, 3.5))
plt.plot(thresholds, f1_curve, label="F1-score")
plt.plot(thresholds, acc_curve, label="Accuracy")
plt.axvline(chosen_threshold, color="gray", linestyle="--", label=f"chosen threshold ({chosen_threshold})")
plt.xlabel("Classification threshold")
plt.title("Out-of-fold performance by threshold")
plt.legend()
plt.show()
```

**What we see:** F1 is flat around its maximum, between thresholds of about 0.40 and 0.50, so the choice is not fragile. Accuracy, in contrast, is highest around a threshold of 0.5 to 0.55. The two metrics want slightly different thresholds, which is why it matters to know which one the competition uses.

### 4.6. Diagnostics of the Chosen Model

We look at the chosen model in more detail on the validation set: how it ranks purchases (ROC curve), what mistakes it makes (confusion matrix), how uncertain the metrics are (bootstrap) and which features matter (permutation importance).

We build the ensemble as a scikit-learn `VotingClassifier`, because it is the object we will retrain in Section 5. First we check that it gives exactly the same probabilities as averaging the four models by hand.

```python
def build_chosen_model():
    """The chosen model as a single scikit-learn object."""
    if chosen_name == "Ensemble (soft voting)":
        return VotingClassifier(
            estimators=[(name, build_model(name)) for name in model_names], voting="soft"
        )
    return build_model(chosen_name)


chosen_model = build_chosen_model().fit(X_tr_f, y_train_array)
chosen_val_proba = chosen_model.predict_proba(X_va_f)[:, 1]

# The object must give the same probabilities as the manual average used in 4.5
assert np.allclose(chosen_val_proba, val_proba[chosen_name])
print("VotingClassifier matches the manual average.")

chosen_val_pred = (chosen_val_proba >= chosen_threshold).astype(int)

# Confusion matrix
cm = confusion_matrix(y_val_array, chosen_val_pred)
display(pd.DataFrame(cm, index=["Actual: no repeat", "Actual: repeat"],
                     columns=["Predicted: no repeat", "Predicted: repeat"]))

# ROC curve
fpr, tpr, _ = roc_curve(y_val_array, chosen_val_proba)
plt.figure(figsize=(4.5, 4.5))
plt.plot(fpr, tpr, label=f"{chosen_name} (AUC = {roc_auc_score(y_val_array, chosen_val_proba):.3f})")
plt.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Baseline (AUC = 0.5)")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title("ROC curve on the validation set")
plt.legend(loc="lower right")
plt.show()
```

**What we see:** the model finds 631 of the 783 purchases that were followed by a repeat purchase (a recall of 81%), but it also predicts "repeat" for 190 of the 515 that were not (37%). The ROC curve is well above the diagonal, which is the baseline. This is the expected behaviour when we choose the threshold to maximise F1.

How reliable are these numbers? We resample the validation set 1,000 times (**bootstrap**) and look at the range of the metrics.

```python
# Bootstrap: resample the validation purchases with replacement and recompute the metrics
boot_rng = np.random.default_rng(SEED)
n_val = len(y_val_array)
boot_auc, boot_f1 = [], []
for _ in range(1000):
    sample = boot_rng.integers(0, n_val, n_val)
    boot_auc.append(roc_auc_score(y_val_array[sample], chosen_val_proba[sample]))
    boot_f1.append(f1_score(y_val_array[sample], chosen_val_pred[sample]))

print("ROC-AUC: %.3f (95%% interval %.3f to %.3f)" % (np.mean(boot_auc), *np.percentile(boot_auc, [2.5, 97.5])))
print("F1     : %.3f (95%% interval %.3f to %.3f)" % (np.mean(boot_f1), *np.percentile(boot_f1, [2.5, 97.5])))
print("Baseline F1 for comparison: %.3f" % baseline_metrics["F1"])
```

**What we see:** the 95% interval of the ROC-AUC goes from about 0.79 to 0.84, and the F1 interval from about 0.76 to 0.81. The lower end of both intervals is above the baseline (ROC-AUC 0.5 and F1 0.7525), so the model really adds information, but the intervals are wide enough that we should not read small differences between models as real.

Finally, which features does the model use? **Permutation importance** shuffles one feature at a time on the validation set and measures how much the ROC-AUC drops. A big drop means the model relies on that feature.

```python
importance = permutation_importance(
    chosen_model, X_va_f, y_val_array, scoring="roc_auc",
    n_repeats=10, random_state=SEED, n_jobs=1,
)
importance_table = pd.DataFrame({
    "Feature": final_features,
    "Mean drop in ROC-AUC": importance.importances_mean,
    "Std": importance.importances_std,
}).sort_values("Mean drop in ROC-AUC", ascending=False)

top = importance_table.head(10)
display(top.round(4).reset_index(drop=True))

ax = top.set_index("Feature")["Mean drop in ROC-AUC"][::-1].plot(
    kind="barh", xerr=top["Std"].to_numpy()[::-1], figsize=(7, 4))
ax.set_title("Permutation importance on the validation set (top 10)")
ax.set_xlabel("Drop in ROC-AUC when the feature is shuffled")
plt.tight_layout()
plt.show()
```

**What we see:** the most important feature by far is `purchase_count_365d` (the number of purchases in the last year), followed by `purchase_count_90d` and `prior_purchase_count`. So the model relies mainly on **how often the customer has bought**, recently and in total. The next features are the commercial region RG17, the line discount, the average gap between purchases and the customer tenure. Most of the other selected features contribute very little on their own, which is normal when several of them carry similar information.

### 4.7. Conclusions and Limitations

- **Result:** a soft-voting ensemble of Logistic Regression, Random Forest, Extra Trees and Gradient Boosting, trained on the 42 selected features, with a threshold chosen by chronological cross-validation. It clearly improves on the baseline (ROC-AUC about 0.82 against 0.5, F1 about 0.79 against 0.75).
- **Models are similar:** the differences between model types are smaller than the uncertainty of the assessment. This is why we choose by robustness (an ensemble with the best cross-validated score) and not by the highest number on the validation set.
- **The threshold matters as much as the model:** moving from 0.5 to the cross-validated threshold increases F1 for every model.

**Limitations:**

1. The validation set was also used in Section 3 to compare feature-selection methods, so its results are slightly optimistic. The 5-fold chronological cross-validation, which did not use it, gives lower numbers (ROC-AUC about 0.79).
2. The test set is from a **later period** than anything we trained on. The repeat rate and the customer mix change over time (for example, there are fewer first purchases in the test set), so performance on the test set may be lower.
3. The threshold was chosen for F1. If the competition uses a different metric, the threshold should be revisited (see the note in 4.1).
